## CELLULE 4 — Reconfigurer et relancer l'entraînement

Cette cellule remplace complètement l'étape 6+7 du notebook précédent.
Elle reconstruit la config et lance l'entraînement avec le bon script.

````python
# ── RECONFIGURATION COMPLÈTE ─────────────────────────────────
import toml, torch, os, time, subprocess, sys, shutil
from pathlib import Path
from google.colab import drive

# ════════════════════════════════════════════════════
# 👇 VÉRIfIE/MODIFIE CES 3 VALEURS
DATASET_DRIVE_PATH = '/content/drive/MyDrive/L_OEIL_DU_KWATT_TRAINING/images'
PROJECT_NAME       = 'L_OEIL_DU_KWATT_TRAINING'
TRIGGER_WORD       = 'shs_cameroon_style'
# ════════════════════════════════════════════════════

# Chemins
BASE_DIR       = Path('/content/lora_training')
SD_SCRIPTS_DIR = BASE_DIR / 'sd-scripts'
DATASET_DIR    = BASE_DIR / 'dataset'
OUTPUT_DIR     = BASE_DIR / 'output'
CONFIG_DIR     = BASE_DIR / 'config'
MODEL_DIR      = BASE_DIR / 'models'
OUTPUT_DRIVE   = f'/content/drive/MyDrive/lora_output_{PROJECT_NAME}'

for d in [DATASET_DIR, OUTPUT_DIR, CONFIG_DIR, MODEL_DIR, BASE_DIR/'logs']:
    d.mkdir(parents=True, exist_ok=True)
Path(OUTPUT_DRIVE).mkdir(parents=True, exist_ok=True)

# ── Trouver le bon script ────────────────────────────────────
print('🔍 Localisation du script d\'entraînement...')

CANDIDATS = [
    SD_SCRIPTS_DIR / 'sdxl_train_network.py',
    SD_SCRIPTS_DIR / 'train_network.py',
    BASE_DIR / 'kohya_ss' / 'sdxl_train_network.py',
    BASE_DIR / 'kohya_ss' / 'train_network.py',
    Path('/content/sd-scripts/sdxl_train_network.py'),
    Path('/content/sd-scripts/train_network.py'),
]

TRAIN_SCRIPT = None
for c in CANDIDATS:
    if c.exists():
        TRAIN_SCRIPT = c
        print(f'   ✅ Trouvé : {TRAIN_SCRIPT}')
        break

if TRAIN_SCRIPT is None:
    print('   ❌ Toujours introuvable — lance d\'abord la CELLULE 3')
    raise SystemExit('Lance la cellule 3 en premier')

# Dossier parent du script (important pour PYTHONPATH)
SCRIPT_DIR = TRAIN_SCRIPT.parent

# ── Monter Drive si pas déjà fait ───────────────────────────
if not Path('/content/drive/MyDrive').exists():
    print('\n🔗 Montage Google Drive...')
    drive.mount('/content/drive')

# ── Vérifier le dataset ──────────────────────────────────────
print('\n📂 Vérification dataset...')
source = Path(DATASET_DRIVE_PATH)
if not source.exists():
    print(f'   ❌ Dataset introuvable : {DATASET_DRIVE_PATH}')
    raise SystemExit('Vérifie DATASET_DRIVE_PATH')

images   = list(source.rglob('*.jpg')) + list(source.rglob('*.png'))
captions = list(source.rglob('*.txt'))
NB_IMAGES = len([i for i in images if i.with_suffix('.txt').exists()])
print(f'   Images avec caption : {NB_IMAGES}')

# ── Paramètres auto ──────────────────────────────────────────
GPU_VRAM   = torch.cuda.get_device_properties(0).total_memory / 1e9
REPEATS    = max(1, round(200 / NB_IMAGES)) if NB_IMAGES < 200 else 1
EPOCHS     = 10
BATCH      = 1
TOTAL_STEPS= (NB_IMAGES * REPEATS * EPOCHS) // BATCH
SAVE_EVERY = min(500, max(100, TOTAL_STEPS // 5))
LR_WARMUP  = int(TOTAL_STEPS * 0.05)

print(f'   GPU VRAM    : {GPU_VRAM:.0f} GB')
print(f'   Total steps : {TOTAL_STEPS}')
print(f'   Save every  : {SAVE_EVERY} steps')

# ── Préparer dataset au format kohya ────────────────────────
print('\n📋 Préparation du dataset format kohya...')
KOHYA_IMG_DIR = DATASET_DIR / f'{REPEATS}_{TRIGGER_WORD}'
KOHYA_IMG_DIR.mkdir(parents=True, exist_ok=True)

# Vérifier si déjà copié
existing = list(KOHYA_IMG_DIR.glob('*.jpg'))
if len(existing) > NB_IMAGES * 0.9:
    print(f'   ✅ Dataset déjà copié ({len(existing)} images) — on saute')
else:
    print(f'   Copie depuis Drive... (peut prendre 2-3 min)')
    copied = 0
    for img in images:
        cap = img.with_suffix('.txt')
        if not cap.exists():
            continue
        rel  = img.relative_to(source)
        name = str(rel).replace('/', '__').replace('\\', '__')
        stem = Path(name).stem
        try:
            shutil.copy2(img, KOHYA_IMG_DIR / f'{stem}.jpg')
            shutil.copy2(cap, KOHYA_IMG_DIR / f'{stem}.txt')
            copied += 1
        except Exception as e:
            pass
    print(f'   ✅ {copied} paires copiées')

# ── Modèle SDXL ─────────────────────────────────────────────
MODEL_PATH = MODEL_DIR / 'sd_xl_base_1.0.safetensors'
if not MODEL_PATH.exists():
    print('\n📥 Téléchargement SDXL (~6.5 GB)...')
    from huggingface_hub import hf_hub_download
    hf_hub_download(
        repo_id='stabilityai/stable-diffusion-xl-base-1.0',
        filename='sd_xl_base_1.0.safetensors',
        local_dir=str(MODEL_DIR),
        local_dir_use_symlinks=False
    )
    print('   ✅ SDXL téléchargé')
else:
    print(f'\n   ✅ SDXL déjà présent ({MODEL_PATH.stat().st_size/1e9:.1f} GB)')

# ── Générer config TOML ──────────────────────────────────────
print('\n⚙️  Génération de la config...')

# Prompts de test générés pendant l'entraînement
sample_prompts = [
    f'{TRIGGER_WORD}, young Cameroonian man, phys_cam_skin, Yaoundé bar interior, vibrant_night_lighting, cinematic --n ugly, blurry --w 768 --h 768 --s 20 --l 7',
    f'{TRIGGER_WORD}, Yaoundé street at night, kwatt_snack, laterite road, neon signs, cinematic --n ugly, western --w 768 --h 768 --s 20 --l 7',
]
PROMPTS_FILE = CONFIG_DIR / 'sample_prompts.txt'
PROMPTS_FILE.write_text('\n'.join(sample_prompts))

# Config d'entraînement
config = {
    'model_arguments': {
        'pretrained_model_name_or_path': str(MODEL_PATH),
        'v2': False,
        'v_parameterization': False,
    },
    'dataset_arguments': {
        'train_data_dir': str(DATASET_DIR),
        'resolution': '768,768',
        'enable_bucket': True,
        'min_bucket_reso': 512,
        'max_bucket_reso': 1024,
        'bucket_reso_steps': 64,
        'bucket_no_upscale': True,
        'caption_extension': '.txt',
        'shuffle_caption': True,
        'keep_tokens': 1,
    },
    'training_arguments': {
        'output_dir': str(OUTPUT_DIR),
        'output_name': PROJECT_NAME,
        'save_model_as': 'safetensors',
        'save_every_n_steps': SAVE_EVERY,
        'save_last_n_steps': 3,
        'max_train_steps': TOTAL_STEPS,
        'max_train_epochs': EPOCHS,
        'train_batch_size': BATCH,
        'gradient_accumulation_steps': 4,
        'gradient_checkpointing': True,
        'mixed_precision': 'fp16',
        'seed': 42,
        'logging_dir': str(BASE_DIR / 'logs'),
        'log_with': 'tensorboard',
        'cache_latents': True,
        'cache_latents_to_disk': False,
        'xformers': True,
        'lowram': GPU_VRAM < 20,
    },
    'optimizer_arguments': {
        'optimizer_type': 'AdamW8bit',
        'learning_rate': 0.0004,
        'unet_lr': 0.0004,
        'text_encoder_lr': 0.0001,
        'lr_scheduler': 'cosine_with_restarts',
        'lr_warmup_steps': LR_WARMUP,
        'lr_scheduler_num_cycles': 3,
    },
    'network_arguments': {
        'network_module': 'networks.lora',
        'network_dim': 32,
        'network_alpha': 16,
    },
    'sample_prompt_arguments': {
        'sample_every_n_steps': SAVE_EVERY,
        'sample_sampler': 'euler_a',
        'sample_prompts': str(PROMPTS_FILE),
    },
}

CONFIG_FILE = CONFIG_DIR / 'training_config.toml'
with open(CONFIG_FILE, 'w') as f:
    toml.dump(config, f)

print(f'   ✅ Config : {CONFIG_FILE}')
print(f'   ✅ Script : {TRAIN_SCRIPT}')
print('\n✅ Tout est prêt — lance la CELLULE 5 pour démarrer')
````

````python
# ── LANCEMENT ENTRAÎNEMENT ───────────────────────────────────
import subprocess, sys, os, time, shutil
from pathlib import Path
from datetime import datetime

print(f'🚀 Démarrage : {datetime.now().strftime("%H:%M:%S")}')
print(f'   Script : {TRAIN_SCRIPT}')
print(f'   Config : {CONFIG_FILE}')
print('─' * 60)

# Commande
cmd = [
    sys.executable,
    str(TRAIN_SCRIPT),
    '--config_file', str(CONFIG_FILE),
]

# Environnement avec PYTHONPATH pointant sur sd-scripts
env = {
    **os.environ,
    'PYTHONPATH': str(SCRIPT_DIR),
    'CUDA_VISIBLE_DEVICES': '0',
}

start_time  = time.time()
loss_values = []

process = subprocess.Popen(
    cmd,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    cwd=str(SCRIPT_DIR),
    env=env
)

for line in process.stdout:
    print(line, end='', flush=True)

    # Extraire la loss pour suivi
    if 'loss=' in line:
        try:
            loss_val = float(line.split('loss=')[1].split(']')[0].split(',')[0].strip())
            loss_values.append(loss_val)
        except:
            pass

    # Copie auto vers Drive à chaque checkpoint
    if 'saving checkpoint' in line.lower() or 'Saving model' in line:
        for ckpt in OUTPUT_DIR.glob('*.safetensors'):
            dest = Path(OUTPUT_DRIVE) / ckpt.name
            if not dest.exists():
                shutil.copy2(ckpt, dest)
                print(f'\n   💾 Drive ← {ckpt.name}')

process.wait()
duration = time.time() - start_time

# Copie finale de tout
print('\n💾 Copie finale vers Drive...')
for f in OUTPUT_DIR.glob('*.safetensors'):
    dest = Path(OUTPUT_DRIVE) / f.name
    shutil.copy2(f, dest)
    print(f'   ✅ {f.name} → Drive')

print('\n' + '═' * 60)
print(f'✅ TERMINÉ en {int(duration//3600)}h{int((duration%3600)//60)}min')
if loss_values:
    print(f'   Loss : {loss_values[0]:.4f} → {loss_values[-1]:.4f}')
    if loss_values[-1] < 0.04:
        print('   ⚠️  Loss très basse → utilise un checkpoint intermédiaire')
    elif loss_values[-1] < 0.12:
        print('   ✅ Loss correcte → bon entraînement')
    else:
        print('   ⚠️  Loss encore haute → augmente epochs ou LR')
print(f'\n   📁 Tes LoRA sont dans Drive : {OUTPUT_DRIVE}/')
print('═' * 60)
````

````python
# ── CELLULE 4 : Solution alternative si cellule 3 échoue ─────
# Patch direct dans le fichier sd-scripts incriminé
# Plus fiable que de changer les versions

from pathlib import Path

print('🔧 Patch direct des fichiers sd-scripts...\n')

SD_SCRIPTS = Path('/content/lora_training/sd-scripts')

# Fichiers qui utilisent cached_download
fichiers_a_patcher = list(SD_SCRIPTS.rglob('*.py'))

patches_appliques = 0
for fichier in fichiers_a_patcher:
    try:
        contenu = fichier.read_text(encoding='utf-8')

        if 'cached_download' not in contenu:
            continue

        # Patch 1 : import cached_download → import hf_hub_download as cached_download
        nouveau = contenu.replace(
            'from huggingface_hub import cached_download',
            'from huggingface_hub import hf_hub_download as cached_download'
        )
        # Patch 2 : variante avec virgule
        nouveau = nouveau.replace(
            'cached_download, hf_hub_download',
            'hf_hub_download'
        )
        nouveau = nouveau.replace(
            'hf_hub_download, cached_download',
            'hf_hub_download'
        )
        # Patch 3 : usage de cached_download() → hf_hub_download()
        nouveau = nouveau.replace(
            'cached_download(',
            'hf_hub_download('
        )

        if nouveau != contenu:
            fichier.write_text(nouveau, encoding='utf-8')
            rel = fichier.relative_to(SD_SCRIPTS)
            print(f'   ✅ Patché : {rel}')
            patches_appliques += 1

    except Exception as e:
        print(f'   ⚠️  {fichier.name} : {e}')

if patches_appliques == 0:
    print('   ℹ️  Aucun fichier avec cached_download trouvé')
    print('   Le problème vient peut-être de diffusers lui-même')
    print()
    print('   Solution finale : désinstaller diffusers système et réinstaller')

    import subprocess, sys
    subprocess.run([sys.executable, '-m', 'pip', 'uninstall', 'diffusers', '-y', '-q'])
    subprocess.run([sys.executable, '-m', 'pip', 'install', 'diffusers==0.25.1', '-q'])
    print('   ✅ diffusers réinstallé en 0.25.1')
else:
    print(f'\n✅ {patches_appliques} fichier(s) patché(s)')

print()
print('━' * 50)
print('👉 Retourne dans correction_kohya_colab.ipynb')
print('   et relance la CELLULE 5 (entraînement)')
print('━' * 50)
````
